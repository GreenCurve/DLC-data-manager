```python
import dlc_manager as dlm
```

```python
project_root_path = r'Absolute\path\for\project\root'
```

```python
prj = dlm.init_data_project(project_root_path)
```

```python
network_project_name = 'My_First_Network'
```

```python
data_folders = ['Choosen_folder_name_1']
# data_folders = ['Choosen_folder_name_1','Choosen_folder_name_2']
```

```python
try:
    net = prj.get_network_project(network_project_name)
except FileNotFoundError:
    net = prj.create_network_project(name=network_project_name)
```

```python
for folder_id in data_folders:
    net.add_labeled_data(prj.frame_set_path(folder_id), overwrite=True)
```

```python
net.create_train_dataset(net_type="resnet_50", num_shuffles=1)
```

```python
net.train_network(epochs=100)
```

```python
net.evaluate_network()
```
