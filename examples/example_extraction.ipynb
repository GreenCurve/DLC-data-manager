```python
import os
import dlc_manager as dlm
```

```python
project_root_path = r'Absolute\path\for\project\root'
```

```python
prj = dlm.init_data_project(project_root_path)
```

```python
cfg = dlm.ExtractionConfig(
    name="dense_sampling",
    algo="kmeans",
    numframes2pick=50,
    start=0.1,
    stop=0.9,
)
```

#### Saving to the store via DataProject

```python
try:
    prj.save_extraction_config(cfg)              # errors if it already exists
except FileExistsError:
    print('Config already exists!')
prj.save_extraction_config(cfg, overwrite=True)  # overwrite an existing one
```

#### Saving directly to the store

```python
store_path = os.path.join(project_root_path,'extraction_configs')
```

```python

try:
    dlm.save_extraction_config(store_path, cfg, overwrite=False)      # errors if it already exists
except FileExistsError:
    print('Config already exists!')
```
